Python Factorial Program Review

**Complete Python Code**

```python
n = 5
fact = 1

for i in range(1, n + 1):
    fact = fact * i

print("Factorial of", n, "=", fact)
```

**Logic:** The program initializes `fact = 1` and multiplies it by every integer from 1 to 5 using a `for` loop.

**Errors:** No logical or syntax errors are present.

**Output:**
```text
Factorial of 5 = 120
```

**Output Verification:** The output is correct because \(5! = 5 \times 4 \times 3 \times 2 \times 1 = 120\).

**Conclusion:** The program correctly calculates and displays the factorial of 5.

**Q2. How can we check the type of a value stored in a Python dictionary?**

We can use the `type()` function to check the data type of a value.

Example:

```python
student = {
    "name": "Rahul",
    "roll": 101,
    "marks": 87.5
}

print(type(student["name"]))
print(type(student["roll"]))
print(type(student["marks"]))
```

**Output:**

```text
<class 'str'>
<class 'int'>
<class 'float'>
```
